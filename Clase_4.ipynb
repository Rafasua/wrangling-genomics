```python
## PORCENTUALES

años = [2019, 2020, 2021, 2022, 2023, 2024, 2025]
cancer_pulmon = pd.Series([22, 35, 55, 45, 47, 42, 39])
cancer_estomago = pd.Series([25, 19, 19, 32, 45, 49, 38])

total = cancer_pulmon + cancer_estomago
P_pulmon = cancer_pulmon / total * 100
P_estomago = cancer_estomago / total * 100

fig = plt.figure(figsize = (6, 5))
ax = fig.add_subplot(111)

ax.bar(años, P_pulmon)
ax.bar(años, P_estomago, bottom = P_pulmon)

plt.show()
```

```python
## PORCENTUALES

años = [2019, 2020, 2021, 2022, 2023, 2024, 2025]
cancer_pulmon = pd.Series([22, 35, 55, 45, 47, 42, 39])
cancer_estomago = pd.Series([25, 19, 19, 32, 45, 49, 38])

total = cancer_pulmon + cancer_estomago
P_pulmon = cancer_pulmon / total * 100
P_estomago = cancer_estomago / total * 100

fig = plt.figure(figsize = (6, 5))
ax = fig.add_subplot(111)

ax.bar(años, P_pulmon, label = "Pulmón")
ax.bar(años, P_estomago, bottom = P_pulmon, label = "Estomago")


for bar in ax.patches:
  ax.text(bar.get_x() + bar.get_width()/2,
          bar.get_y() + bar.get_height()/2,
          round(bar.get_height(), 1),
          ha = "center", color = "white")

plt.legend(bbox_to_anchor = (1, 0.6))
plt.show()
```

```python
df = pd.read_csv("COVID-19 Pandemic in Peru.csv")
df
```

```python
from matplotlib.ticker import AutoMinorLocator

fig = plt.figure(figsize = (10, 8))
ax = fig.add_subplot(111)

# Prop
prop = df.new_cases.max() / df.people_vaccinated.max()

# Gráfico de Lineas
ax.plot(df.index, df.new_cases/1000, color = "#32408C",
        label = "Daily cases\nper thousand")

# Grafico de Area
ax.fill_between(df.index, df.people_vaccinated*prop/1000, color = "#F2DFEB",
                label = "Cumulative people\nvaccinated per million")

# Crear segundo eje

secax = ax.secondary_yaxis("right", functions = (lambda x: x/prop/1000, lambda x: x/prop/1000))
secax.set_ylabel("Cumulative people\nvaccinated per million", size = 15, labelpad = 10)

# Ejes x e y
ax.set_ylabel("Daily cases\nper thousand", size = 15, labelpad = 10)
ax.set_xlabel("Date", size = 15)

# Establecer las fechas
fechas = ["2020-03-06", "2021-03-06", "2022-03-06", "2023-03-06"]
ax.set_xticks(np.arange(0, len(df.index), 365))
ax.set_xticklabels(fechas)

ax.set_yticks(np.arange(0, 61, 20))
secax.set_yticks(np.arange(0, 31, 10))

# Cuadricula
ax.xaxis.set_minor_locator(AutoMinorLocator(4))
ax.yaxis.set_minor_locator(AutoMinorLocator(2))

plt.grid(which = "both", linestyle = "dashed", linewidth = 1.5)

# Espinas
ax.spines["top"].set_color("none")

plt.legend(bbox_to_anchor = (1.4, 0.6))

plt.show()
```
